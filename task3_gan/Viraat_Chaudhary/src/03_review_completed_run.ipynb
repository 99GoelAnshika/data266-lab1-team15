# Completed-run evidence review — CPU only
This notebook reads the preserved run. It does not execute training or inference.
Its outputs describe the original GPU run; the review itself executes on CPU.


```python
from pathlib import Path
import csv, json
MEMBER = Path.cwd().parent if Path.cwd().name == 'src' else Path.cwd()
if not (MEMBER / 'logs').exists():
    MEMBER = Path.cwd() / 'task3_gan' / 'Viraat_Chaudhary'
RUN = 'viraat_resizeconv6_run001'
events = [json.loads(line) for line in (MEMBER / 'logs' / RUN / 'events.jsonl').read_text().splitlines() if line]
completed = [e for e in events if e['event_type'] == 'epoch_end']
print('Completed epochs:', len(completed))
print('Final completion record:', json.dumps(events[-1], indent=2))
assert len(completed) == 60 and events[-1]['event_type'] == 'run_complete'
```

```python
metrics = MEMBER / 'outputs' / RUN / 'metrics'
training = json.loads((metrics / 'training_summary.json').read_text())
evaluation = json.loads((metrics / 'evaluation_summary.json').read_text())
print('Recorded training:', json.dumps(training, indent=2))
print('Saved evaluation:', json.dumps(evaluation, indent=2))
print('Professor CSV:')
print((MEMBER / 'submission.csv').read_text())
```

```python
audit = MEMBER / 'outputs' / RUN / 'human_audit'
for number in [1, 2]:
    with (audit / f'rater{number}_scores.csv').open() as handle:
        rows = list(csv.DictReader(handle))
    filled = sum(all((r.get(c) or '').strip() in ['1','2','3','4','5'] for c in ['style','content','artifacts']) for r in rows)
    print(f'Rater {number}: {filled}/{len(rows)} complete rows')
print('No human scores or current Kaggle score are invented by this review.')
```
